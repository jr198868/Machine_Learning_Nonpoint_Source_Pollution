```python
import pandas as pd
import csv
import os
import shutil
import matplotlib.pyplot as plt
from PIL import Image
```

```python

# Define source directories
prefex_list = [
    "/Users/rjing/Desktop/Machine_Learning_Nonpoint_Source_Pollution/data/water_runoff_sample/Total_Phosphorus/",
    "/Users/rjing/Desktop/Machine_Learning_Nonpoint_Source_Pollution/data/water_runoff_sample/Total_Nitrogen/",
    "/Users/rjing/Desktop/Machine_Learning_Nonpoint_Source_Pollution/data/water_runoff_sample/Ammoniacal_Nitrogen/",
    "/Users/rjing/Desktop/Machine_Learning_Nonpoint_Source_Pollution/data/water_runoff_sample/Dissolved_Phosphorus/",
    "/Users/rjing/Desktop/Machine_Learning_Nonpoint_Source_Pollution/data/water_runoff_sample/Nitrate_Nitrogen/",
    "/Users/rjing/Desktop/Machine_Learning_Nonpoint_Source_Pollution/data/water_runoff_sample/Particulate_Phosphorus/"
]

# Destination folder
dest_folder = "/Users/rjing/Desktop/Machine_Learning_Nonpoint_Source_Pollution/data/weather_Conditions_Across_Clusters_plots/"
os.makedirs(dest_folder, exist_ok=True)

# Move PNG files from subfolders to the destination folder
for prefex in prefex_list:
    scatter_plots_dir = os.path.join(prefex, "weather_Conditions_Across_Clusters")
    if os.path.exists(scatter_plots_dir):  # Check if scatter_plots folder exists
        for filename in os.listdir(scatter_plots_dir):
            if filename.endswith(".png"):
                src_path = os.path.join(scatter_plots_dir, filename)
                dest_path = os.path.join(dest_folder, filename)
                shutil.move(src_path, dest_path)  # Move the file

print(f"Moved all scatter plot images to {dest_folder}")
```

```python

# Destination folder containing PNG images
dest_folder = "/Users/rjing/Desktop/Machine_Learning_Nonpoint_Source_Pollution/data/cluster_scatter_plots/"

# List all PNG files in the directory
file_list = [f for f in os.listdir(dest_folder) if f.endswith(".png")]
file_list.sort()  # Sort files for consistent order

# Load images
images = [Image.open(os.path.join(dest_folder, f)) for f in file_list]

# Define grid size (6 rows × 4 columns)
rows, cols = 6, 4
total_slots = rows * cols

# Create figure with larger subplots
fig, axes = plt.subplots(rows, cols, figsize=(cols * 5, rows * 5))  # Bigger subfigures

# Reduce spacing between plots
plt.subplots_adjust(wspace=0.05, hspace=0.05)  # Minimize space between plots

# Flatten axes array for easy indexing
axes = axes.flatten()

# Add images to subplots
for i, img in enumerate(images[:total_slots]):  # Limit to first 24 images
    axes[i].imshow(img)
    axes[i].axis("off")  # Hide axes

# Remove empty subplots if fewer than 24 images
for j in range(len(images), total_slots):
    fig.delaxes(axes[j])

# Save as PDF with higher resolution
output_pdf = "/Users/rjing/Desktop/combined_scatter_plots.pdf"
plt.savefig(output_pdf, format="pdf", bbox_inches="tight", dpi=400)  # Higher DPI for better quality

print(f"Saved combined figure as {output_pdf}")
```
